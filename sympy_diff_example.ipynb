```python
import sympy as sp
from sympy import Matrix, cos, sin
```

```python
x = sp.symbols("x1, x2")
```

```python

x_eq = Matrix( [
                    [cos(x[0]) + cos(x[1]), sin(x[0]) + cos(x[1])],
                    [sin(x[0]) - cos(x[1]), sin(x[0]) + sin(x[1])]
                ])
```

```python
print("The variable angles")
print(x)
```

```python
print("Equation Matrix")
print(x_eq)
```

```python
print("Differentiate Equation matrix w.r.t. x1 : ")
print(sp.diff(x_eq, x[0]))
```
